## Test olmOCR

```python
## import dependencies
import torch
from transformers import AutoProcessor, Qwen2_5_VLForConditionalGeneration
from olmocr.prompts import build_no_anchoring_v4_yaml_prompt
from PIL import Image
import os
import time
```

```python
# --- 1. Configure Paths ---
# IMPORTANT: Update this to the actual path where you saved the 4-bit models folder
model_directory = "./models/olmOCR-2-7B-4bit"

# Check if paths exist
if not os.path.isdir(model_directory):
    raise FileNotFoundError(f"Model directory not found: {model_directory}. Please update the path.")
```

```python
# --- 2. Check for GPU ---
if not torch.cuda.is_available():
    print("WARNING: No GPU detected. This models requires a GPU to run efficiently (or at all).")
    # Attempt to run on CPU, but it will be extremely slow or might fail
    device = torch.device("cpu")
    device_map_config = "cpu"
else:
    device = torch.device("cuda")
    device_map_config = "auto" # Let accelerate handle GPU placement
    print(f"GPU detected: {torch.cuda.get_device_name(0)}")
```

```python
# --- 3. Load Model and Processor ---
print(f"Loading processor from {model_directory}...")
# Use the correct processor for Qwen2.5-VL
processor = AutoProcessor.from_pretrained("Qwen/Qwen2-VL-7B-Instruct")

print(f"Loading 4-bit models from {model_directory}...")
# When loading a models saved with BitsAndBytesConfig,
# `from_pretrained` with `device_map='auto'` should automatically
# load it in 4-bit on the GPU.
try:
    model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
        model_directory,
        device_map = "cuda", # Use 'auto' for GPU, 'cpu' otherwise
        torch_dtype=torch.bfloat16 # Compute dtype, adjust if needed
    )
except Exception as e:
    print(f"\n--- ERROR loading models ---")
    print(f"Ensure you have 'bitsandbytes' installed and a compatible GPU environment.")
    print(f"If running on CPU was intended, it might fail due to quantization requirements.")
    print(f"Original error: {e}")
    raise

print("Model and processor loaded successfully.")
```

```python
# IMPORTANT: Update this to the path of the image you want to process
local_image_path = "./data/dummy-aadhaar.png"

if not os.path.isfile(local_image_path):
     raise FileNotFoundError(f"Local image file not found: {local_image_path}. Please update the path.")
```

```python
# --- 4. Load and Prepare Image ---
print(f"Loading image: {local_image_path}...")
try:
    input_image = Image.open(local_image_path).convert("RGB")
except Exception as e:
    print(f"Error opening image file: {e}")
    raise
```

```python
# --- 5. Build Prompt and Process Inputs ---
print("Building prompt...")
# Note: For local images, we pass the PIL image directly to the processor
messages = [
    {
        "role": "user",
        "content": [
            {"type": "text", "text": build_no_anchoring_v4_yaml_prompt()},
            {"type": "image"}, # Placeholder for the image
        ],
    }
]

# Process inputs using the processor
text_prompt = processor.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
inputs = processor(
    text=[text_prompt],
    images=[input_image], # Pass the loaded PIL image here
    padding=True,
    return_tensors="pt",
).to(model.device) # Move inputs to the same device as the models
```

```python
# --- 6. Generate Output ---
print("Generating OCR text (this might take a moment)...")
start_time = time.time()
with torch.no_grad(): # Ensure gradients aren't calculated
     output = model.generate(**inputs, max_new_tokens=2048) # Increase max_new_tokens if needed
end_time = time.time()
print(f"Inference took {end_time - start_time:.2f} seconds.")
```

```python
# --- 7. Decode and Print ---
prompt_length = inputs["input_ids"].shape[1]
generated_tokens = output[:, prompt_length:]
generated_text = processor.tokenizer.batch_decode(generated_tokens, skip_special_tokens=True)[0]

print("\n--- OCR Output ---")
print(generated_text)
print("------------------")
```
