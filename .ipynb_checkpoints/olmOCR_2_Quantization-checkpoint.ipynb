```python
import torch
from transformers import AutoProcessor, Qwen2_5_VLForConditionalGeneration, BitsAndBytesConfig
```

```python
# 1. Define the models ID and the directory to save the 4-bit models
base_model_id = "allenai/olmOCR-2-7B-1025"
quantized_model_dir = "./olmOCR-2-7B-4bit"
```

```python
# 2. Configure 4-bit quantization
quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_use_double_quant=True,
)

model_4bit = AutoModelForCausalLM.from_pretrained(model_id, quantization_config=bnb_config_4bit, low_cpu_mem_usage=True)

print(f"Loading base model: {base_model_id}")
```

```python
# 3. Load the models and quantize it to 4-bit
model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
    base_model_id,
    quantization_config=quantization_config,
    device_map="auto"  # Automatically uses the GPU
)
```

```python
# 4. Load the processor (pre-processes images and text)
# The Qwen2.5-VL-7B-Instruct processor is the correct one, as per the models card
processor = AutoProcessor.from_pretrained("Qwen/Qwen2.5-VL-7B-Instruct")

print(f"Saving 4-bit quantized model to: {quantized_model_dir}")
```

```python
# 5. Save the 4-bit models and the processor
model.save_pretrained(quantized_model_dir)
processor.save_pretrained(quantized_model_dir)

print("Quantization and saving complete!")
```
